```python
import numpy as np
import matplotlib.pyplot as plt
```

```python


# Classes
classes = [
    "Background",
    "Low-Mod Risk",
    "High Risk",
    "Very High Risk",
    "Extreme Risk"
]

# Original categorical crossentropy results
original_precision = [0.9058511579, 0.6219546594, 0.6098491281, 0.6649849509, 0.8209918215]
original_recall    = [0.9686229197, 0.4765062547, 0.5001392638, 0.5076970600, 0.4732269300]
original_f1        = [0.9361859987, 0.5396009663, 0.5495724031, 0.5757927578, 0.6003860457]

# Weighted categorical crossentropy results
weighted_precision = [0.9046558546, 0.6458637341, 0.6939090122, 0.7362773340, 0.8545546202]
weighted_recall    = [0.9634046695, 0.5656960447, 0.5342760423, 0.5637340677, 0.5406858805]
weighted_f1        = [0.9331079157, 0.6031275818, 0.6037184045, 0.6385553479, 0.6623168078]

metrics = {
    "Precision": (original_precision, weighted_precision),
    "Recall": (original_recall, weighted_recall),
    "F1 Score": (original_f1, weighted_f1)
}

x = np.arange(len(classes))
bar_width = 0.35

fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

for ax, (metric_name, (original_vals, weighted_vals)) in zip(axes, metrics.items()):
    ax.bar(x - bar_width / 2, original_vals, bar_width, label="Original CCE")
    ax.bar(x + bar_width / 2, weighted_vals, bar_width, label="Weighted CCE")

    ax.set_title(metric_name)
    ax.set_xticks(x)
    ax.set_xticklabels(classes, rotation=35, ha="right")
    ax.set_ylim(0, 1.05)
    ax.set_ylabel("Score")
    ax.grid(axis="y", alpha=0.3)

    # Add value labels above bars
    for i, value in enumerate(original_vals):
        ax.text(i - bar_width / 2, value + 0.02, f"{value:.2f}",
                ha="center", va="bottom", fontsize=8)

    for i, value in enumerate(weighted_vals):
        ax.text(i + bar_width / 2, value + 0.02, f"{value:.2f}",
                ha="center", va="bottom", fontsize=8)

axes[0].legend()

plt.suptitle("Original Categorical Crossentropy vs Weighted Categorical Crossentropy, Pala Mesa Dataset", fontsize=16)
plt.tight_layout()
plt.show()
```
