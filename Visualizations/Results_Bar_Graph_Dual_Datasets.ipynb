```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
classes = [
    "Background",
    "Low-Mod Risk",
    "High Risk",
    "Very High Risk",
    "Extreme Risk"
]

# main_dual_datasets.ipynb
dual_dataset_f1 = [
    0.8722156085,
    0.0355667800,
    0.3632922214,
    0.0873441427,
    0.4388498315
]

# Dual_Datasets_Weighted_Loss_Results_Doc.pdf
dual_weighted_f1 = [
    0.8816991793,
    0.3667867968,
    0.2782057305,
    0.1923680394,
    0.2612219564
]

x = np.arange(len(classes))
bar_width = 0.35

plt.figure(figsize=(10, 6))

plt.bar(x - bar_width / 2, dual_dataset_f1, bar_width, label="Dual Dataset")
plt.bar(x + bar_width / 2, dual_weighted_f1, bar_width, label="Dual Dataset + Weighted Loss")

plt.title("F1 Score: Dual Dataset vs Dual Dataset + Weighted Loss")
plt.ylabel("F1 Score")
plt.ylim(0, 1.05)
plt.xticks(x, classes, rotation=30, ha="right")
plt.legend()
plt.grid(axis="y", alpha=0.3)

for i, value in enumerate(dual_dataset_f1):
    plt.text(i - bar_width / 2, value + 0.02, f"{value:.2f}", ha="center", fontsize=9)

for i, value in enumerate(dual_weighted_f1):
    plt.text(i + bar_width / 2, value + 0.02, f"{value:.2f}", ha="center", fontsize=9)

plt.tight_layout()
plt.show()
```
