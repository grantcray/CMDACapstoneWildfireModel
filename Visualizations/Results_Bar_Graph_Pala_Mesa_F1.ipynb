```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Classes
classes = [
    "Background",
    "Low-Mod Risk",
    "High Risk",
    "Very High Risk",
    "Extreme Risk"
]

# F1 scores only
original_f1 = [0.9361859987, 0.5396009663, 0.5495724031, 0.5757927578, 0.6003860457]
weighted_f1 = [0.9331079157, 0.6031275818, 0.6037184045, 0.6385553479, 0.6623168078]

x = np.arange(len(classes))
bar_width = 0.35

# Create single plot
fig, ax = plt.subplots(figsize=(10, 6))

# Bars
ax.bar(x - bar_width / 2, original_f1, bar_width, label="Original Loss Function")
ax.bar(x + bar_width / 2, weighted_f1, bar_width, label="Weighted Loss Function")

# Formatting
ax.set_title("F1 Score Comparison")
ax.set_xticks(x)
ax.set_xticklabels(classes, rotation=35, ha="right")
ax.set_ylim(0, 1.05)
ax.set_ylabel("F1 Score")
ax.grid(axis="y", alpha=0.3)

# Add value labels
for i, value in enumerate(original_f1):
    ax.text(i - bar_width / 2, value + 0.02, f"{value:.2f}",
            ha="center", va="bottom", fontsize=8)

for i, value in enumerate(weighted_f1):
    ax.text(i + bar_width / 2, value + 0.02, f"{value:.2f}",
            ha="center", va="bottom", fontsize=8)

# Legend and overall title
ax.legend()

plt.suptitle(
    "Original vs Weighted Loss Function F1 Scores, Pala Mesa Dataset",
    fontsize=16
)

plt.tight_layout()
plt.show()
```
